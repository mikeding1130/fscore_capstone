## F-score (Piotroski 2000) — two layers

**Layer 1 — raw numbers.** Ratios and year-on-year differences, in the firm's own
units. Nothing is thresholded here, so every number stays inspectable: a signal that
looks wrong can be traced back to the two statement lines it came from.

**Layer 2 — points.** One comparator per raw number turns it into a 0/1 indicator;
the nine indicators sum to the F-score.

| signal | raw (layer 1) | point (layer 2) |
|---|---|---|
| `roa` | `net_income_parent / total_assets[t-1]` | `> 0` |
| `cfoa` | `cfo / total_assets[t-1]` | `> 0` |
| `d_roa` | `roa[t] - roa[t-1]` | `> 0` |
| `accrual` | `roa - cfoa` | `< 0` (cash beats earnings) |
| `d_lever` | `lever[t] - lever[t-1]`, `lever = long_term_debt / mean(total_assets[t], total_assets[t-1])` | `< 0` |
| `d_liquid` | `liquid[t] - liquid[t-1]`, `liquid = current_assets / current_liabilities` | `> 0` |
| `eq_offer` | equity raised during the year (VND) | `<= 0` (did not issue) |
| `d_margin` | `margin[t] - margin[t-1]`, `margin = gross_profit / net_sales` | `> 0` |
| `d_turn` | `turn[t] - turn[t-1]`, `turn = net_sales / total_assets[t-1]` | `> 0` |

Scaling follows the paper: **beginning-of-year** total assets for `roa`, `cfoa` and
`turn`, **average** total assets for `lever`. Both need `t-1`, so a symbol's first
year in the panel scores nothing — and `d_roa`, `d_lever` and `d_turn` need `t-2`,
because the level they difference is itself scaled by the prior year's assets.

**NA is not zero.** A missing raw number yields `<NA>` for that indicator, never a
lost point. `f_score` is only formed where all nine are available (`min_signals`);
`signals_available` says how many there were. Silently scoring a firm 4/9 because
five inputs were missing would put it in the same bucket as a genuinely weak firm.

### Layer 1 — raw numbers

Two steps: the level ratios for year `t`, then the differences against the same
symbol's previous year.

Lags are taken **inside this panel**, which is already restricted to firm-years that
passed the accounting checks. A year rejected there is a hole here, and `contiguous`
is what stops `t` being differenced against `t-2` as if nothing were missing.

```python
def _div(num: pd.Series, den: pd.Series) -> pd.Series:
    """num/den with a zero denominator yielding NaN rather than inf."""
    return num / den.replace(0, np.nan)


def contiguity(panel: pd.DataFrame) -> pd.Series:
    """True where the previous row of the same symbol is exactly one year back."""
    year = pd.Series(panel.index.get_level_values("period"), index=panel.index).astype("float")
    return year.sub(year.groupby(level="symbol", sort=False).shift(1)).eq(1)


def _prev(s: pd.Series, contiguous: pd.Series) -> pd.Series:
    """Previous row of the same symbol, NaN where that row is not exactly one year back."""
    return s.groupby(level="symbol", sort=False).shift(1).where(contiguous)
```

```python
def ratio_levels(panel: pd.DataFrame) -> pd.DataFrame:
    """
    The level ratios the nine signals are built from, one row per firm-year.

    Kept as its own frame because five of the nine signals are differences of these
    levels: computing the level once and lagging it is what makes ROA[t-1] scale by
    total assets at t-2, as the paper defines it, instead of quietly reusing t-1.
    """
    p = panel.sort_index()
    contiguous = contiguity(p)

    ta = p["total_assets"]
    ta_begin = _prev(ta, contiguous)              # beginning-of-year assets
    ta_avg = (ta + ta_begin) / 2

    # Margin and turnover off zero or negative revenue are not small numbers, they are
    # meaningless ones. NaN here so layer 2 reports NA instead of scoring the noise.
    sales = p["net_sales"].where(p["net_sales"] > 0)

    lv = pd.DataFrame(index=p.index)
    lv["contiguous"] = contiguous
    lv["total_assets_begin"] = ta_begin
    lv["roa"] = _div(p["net_income_parent"], ta_begin)
    lv["cfoa"] = _div(p["cfo"], ta_begin)
    lv["lever"] = _div(p["long_term_debt"], ta_avg)
    lv["liquid"] = _div(p["current_assets"], p["current_liabilities"])
    lv["margin"] = _div(p["gross_profit"], sales)
    lv["turn"] = _div(sales, ta_begin)
    lv["equity_raised"] = p["stock_issuance_proceeds"]
    return lv
```

```python
levels = ratio_levels(df)
levels
```

```python
# Levels that are differenced year-on-year. Lagging the level (rather than lagging the
# inputs and recomputing) is what propagates NaN correctly: d_roa needs three
# consecutive years, and roa[t-1] is already NaN when t-2 is missing.
LAGGED_LEVELS = ("roa", "lever", "liquid", "margin", "turn")


def raw_layer(levels: pd.DataFrame) -> pd.DataFrame:
    """The nine raw numbers, same order as FSCORE_INPUTS. No thresholds applied."""
    contiguous = levels["contiguous"]
    prev = {k: _prev(levels[k], contiguous) for k in LAGGED_LEVELS}

    raw = pd.DataFrame(index=levels.index)
    raw["roa"] = levels["roa"]
    raw["cfoa"] = levels["cfoa"]
    raw["d_roa"] = levels["roa"] - prev["roa"]
    raw["accrual"] = levels["roa"] - levels["cfoa"]
    raw["d_lever"] = levels["lever"] - prev["lever"]
    raw["d_liquid"] = levels["liquid"] - prev["liquid"]
    raw["eq_offer"] = levels["equity_raised"]
    raw["d_margin"] = levels["margin"] - prev["margin"]
    raw["d_turn"] = levels["turn"] - prev["turn"]
    return raw
```

```python
raw = raw_layer(levels)
raw
```

### Layer 2 — points

One comparator per raw number. Each rule reads only its own raw column, which is what
keeps the NA rule uniform: an indicator is NA exactly when its raw number is NaN.

```python
@dataclass(frozen=True)
class Signal:
    name: str                                    # matches an FSCORE_INPUTS key
    test: Callable[[pd.Series], pd.Series]       # raw number -> point earned
    note: str


SIGNALS = (
    Signal("roa", lambda x: x > 0,
           "Profitable on beginning-of-year assets."),
    Signal("cfoa", lambda x: x > 0,
           "Operations threw off cash."),
    Signal("d_roa", lambda x: x > 0,
           "Profitability improved. Both years scale by their own beginning assets, "
           "so this needs three consecutive years, not two."),
    Signal("accrual", lambda x: x < 0,
           "CFO exceeds net income. Earnings backed by cash rather than by accruals; "
           "the paper's quality-of-earnings signal."),
    Signal("d_lever", lambda x: x < 0,
           "Long-term leverage fell. Strictly as published: a firm with no long-term "
           "debt in either year has d_lever == 0 and earns nothing here."),
    Signal("d_liquid", lambda x: x > 0,
           "Current ratio improved."),
    Signal("eq_offer", lambda x: x <= 0,
           "Did not raise equity. Issuing shares is read as a signal that internal "
           "funds could not cover the firm's needs."),
    Signal("d_margin", lambda x: x > 0,
           "Gross margin improved."),
    Signal("d_turn", lambda x: x > 0,
           "Asset turnover improved."),
)

# The signal set and FSCORE_INPUTS must not drift apart: FSCORE_INPUTS is what decides
# which statement lines get pulled, so a signal missing there would be scored off
# columns nobody guaranteed were fetched.
assert [s.name for s in SIGNALS] == list(FSCORE_INPUTS), "SIGNALS drifted from FSCORE_INPUTS"

SIGNAL_COLS = [f"f_{s.name}" for s in SIGNALS]


def signals_table() -> pd.DataFrame:
    """The rule set as a frame — for review, export, or the appendix."""
    return pd.DataFrame([{
        "signal": s.name,
        "inputs": ", ".join(k for _, k in FSCORE_INPUTS[s.name]),
        "note": " ".join(s.note.split()),
    } for s in SIGNALS]).set_index("signal")


signals_table()
```

```python
def score_layer(raw: pd.DataFrame, signals=SIGNALS, min_signals: int = 9) -> pd.DataFrame:
    """
    Raw numbers -> f_<signal> indicators + f_score.

    True = point earned, False = not earned, NA = the raw number was missing.
    f_score is the sum of the indicators, and is NA below min_signals available:
    a partial sum is not a low score, and mixing the two would put firms with holes
    in their statements into the same bucket as genuinely weak ones.
    """
    out = {}
    for s in signals:
        point = s.test(raw[s.name]).astype("boolean")
        point[raw[s.name].isna()] = pd.NA
        out[f"f_{s.name}"] = point

    scores = pd.DataFrame(out, index=raw.index)
    available = scores.notna().sum(axis=1)
    total = scores.fillna(False).sum(axis=1).astype("Int64")

    scores["signals_available"] = available
    scores["f_score"] = total.where(available >= min_signals)
    return scores
```

```python
scores = score_layer(raw)
scores
```

### Coverage and diagnostics

Read these before using the score. The first table is the one that decides how much of
the panel is actually scoreable; the rest are sanity checks on the nine signals.

```python
coverage = pd.DataFrame({
    "rows": scores["signals_available"].value_counts().sort_index(),
})
coverage["pct"] = (100 * coverage["rows"] / len(scores)).round(2)
coverage.index.name = "signals_available"
coverage
```

```python
pd.DataFrame({
    "earned": (scores[SIGNAL_COLS] == True).sum(),      # noqa: E712
    "not_earned": (scores[SIGNAL_COLS] == False).sum(),  # noqa: E712
    "not_applicable": scores[SIGNAL_COLS].isna().sum(),
}).assign(earned_pct=lambda d: (100 * d["earned"] / (d["earned"] + d["not_earned"])).round(2))
```

```python
dist = scores["f_score"].value_counts(dropna=False).sort_index().rename("rows").to_frame()
dist["pct"] = (100 * dist["rows"] / len(scores)).round(2)
dist
```

```python
scores.groupby(level="period")["f_score"].agg(
    scored="count", mean="mean", high=lambda s: (s >= 8).sum(), low=lambda s: (s <= 1).sum()
).round(2)
```

```python
# d_lever is the one signal whose zero point is not a judgement about the firm: a
# debt-free firm cannot deleverage, so the published `< 0` rule denies it the point.
#
# d_lever == 0 says the RATIO did not move, which is a weaker claim than "no debt" —
# so the two are counted apart below rather than assumed equal. They coincide in this
# panel for a structural reason: lever = long_term_debt / mean(total_assets[t],
# total_assets[t-1]), so a firm carrying debt would need both years' ratios to agree to
# the last float bit, which average total assets never does. Firms that repaid debt land
# in d_lever < 0 and earn the point; firms that borrowed land in d_lever > 0.
ltd = df["long_term_debt"].sort_index()
ltd_prev = _prev(ltd, levels["contiguous"])

scored = scores["f_score"].notna()
zero = scored & (raw["d_lever"] == 0)

pd.Series({
    "scored_rows": int(scored.sum()),
    "d_lever_negative": int((scored & (raw["d_lever"] < 0)).sum()),
    "d_lever_positive": int((scored & (raw["d_lever"] > 0)).sum()),
    "d_lever_zero": int(zero.sum()),
    "  zero, no debt in either year": int((zero & (ltd == 0) & (ltd_prev == 0)).sum()),
    "  zero, debt outstanding": int((zero & ((ltd > 0) | (ltd_prev > 0))).sum()),
    "mean_f_score": round(scores.loc[scored, "f_score"].mean(), 3),
    "mean_f_score_if_rule_were_le_0": round(
        (scores.loc[scored, "f_score"] + (raw.loc[scored, "d_lever"] == 0)).mean(), 3),
})
```

```python
# How many names the strategy actually buys under each d_lever rule, at the cutoff in
# use. This is the number that matters: the mean F-score barely moves, but the rule
# decides how many firm-years clear the bar.
#
# Loosening `< 0` to `<= 0` can only ADD points, so the strict selection is a subset of
# the loose one — every difference below is a firm-year crossing the cutoff, none
# leaving it. The assert is what guarantees that reading of the table.
from dataclasses import replace

CUTOFF = 7

SIGNALS_LEVER_LE0 = tuple(replace(s, test=lambda x: x <= 0) if s.name == "d_lever" else s
                          for s in SIGNALS)
scores_le0 = score_layer(raw, signals=SIGNALS_LEVER_LE0)

selected = (scores["f_score"] >= CUTOFF).fillna(False)
selected_le0 = (scores_le0["f_score"] >= CUTOFF).fillna(False)
assert not (selected & ~selected_le0).any(), "loosening dropped a name — rules are not nested"

cutoff_shift = pd.DataFrame({
    "scored": scores["f_score"].notna().groupby(level="period").sum(),
    "d_lever<0": selected.groupby(level="period").sum(),
    "d_lever<=0": selected_le0.groupby(level="period").sum(),
})
cutoff_shift["added"] = cutoff_shift["d_lever<=0"] - cutoff_shift["d_lever<0"]
cutoff_shift["added_pct"] = (100 * cutoff_shift["added"] / cutoff_shift["d_lever<0"]).round(1)

cutoff_shift.loc["ALL"] = cutoff_shift.sum()
cutoff_shift.loc["ALL", "added_pct"] = round(
    100 * cutoff_shift.loc["ALL", "added"] / cutoff_shift.loc["ALL", "d_lever<0"], 1)

# The ALL row makes the index object-typed, which floats the counts. They are names
# held, not quantities: keep them integers.
counts = ["scored", "d_lever<0", "d_lever<=0", "added"]
cutoff_shift[counts] = cutoff_shift[counts].astype(int)
cutoff_shift
```
