```python
import pandas as pd
import sqlite3

con = sqlite3.connect('../cafef.db')
```

```python
pd.read_sql("SELECT * from fireant_prices where date <= '2025-12-31' order by date", con)
```

```python
price_df = pd.read_sql("SELECT symbol, count(*) as price_available_day from fireant_prices where date <= '2025-12-31' group by symbol", con)
price_df.set_index("symbol", inplace=True)
price_df
```

```python
financial_data_coverage_grid = pd.read_csv("../data/fireant_grid_sorted.csv")
financial_data_coverage_grid.rename({"Unnamed: 0": "symbol", "Unnamed: 0.1": "ID"}, axis=1, inplace=True)
financial_data_coverage_grid.set_index("symbol", inplace=True)
financial_data_coverage_grid
```

```python
financial_data_coverage_grid.join(price_df, on="symbol", how="left")
```

```python
import numpy as np
year_cols = [col for col in financial_data_coverage_grid.columns if col.isdigit()]
financial_data_coverage_grid['coverage_score_year'] = financial_data_coverage_grid[year_cols].clip(upper=3).sum(axis=1)/3
financial_data_coverage_grid
```

```python
financial_data_coverage_grid = financial_data_coverage_grid.join(price_df, on="symbol", how="left")
financial_data_coverage_grid["price_available_day"] = financial_data_coverage_grid["price_available_day"].fillna(0)
financial_data_coverage_grid
```

```python
financial_data_coverage_grid.query("~(coverage_score_year < 2 or price_available_day == 0)")
```

```python
financial_data_coverage_grid.query("~(coverage_score_year < 2 or price_available_day == 0)").to_csv("../data/preprocessing_pipeline_results/simple_filter_fa_price.csv")

financial_data_coverage_grid.query("(coverage_score_year < 2 or price_available_day == 0)").to_csv("../data/preprocessing_pipeline_results/simple_filter_fa_price_holdout.csv")
```
